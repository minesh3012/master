```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

# Project paths
PROJECT_ROOT = Path.cwd().parents[0] if "notebooks" in str(Path.cwd()) else Path.cwd()
DATA_RAW_DIR = PROJECT_ROOT / "data" / "raw"

DATA_RAW_DIR
```

```python
raw_files = list(DATA_RAW_DIR.rglob("*.csv"))
raw_files
```

```python
if not raw_files:
    raise FileNotFoundError("No raw CSV files found in data/raw/. Please add your NT export.")

df = pd.read_csv(raw_files[0])
df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
df.isna().sum().sort_values(ascending=False)
```

```python
if "Time" in df.columns:
    df["Time"] = pd.to_datetime(df["Time"], errors="coerce")
    df["Time"].head()
else:
    print("No 'Time' column found.")
```

```python
plt.figure(figsize=(14,5))
plt.plot(df["Time"], df["Close"], linewidth=0.8)
plt.title("Close Price Over Time")
plt.xlabel("Time")
plt.ylabel("Close")
plt.grid(True)
plt.show()
```

```python
if "Range" in df.columns:
    plt.figure(figsize=(10,4))
    sns.histplot(df["Range"], bins=100, kde=True)
    plt.title("Distribution of Candle Range")
    plt.show()
else:
    print("Column 'Range' not found.")
```

```python
if "Vol" in df.columns:
    plt.figure(figsize=(10,4))
    sns.histplot(df["Vol"], bins=100, kde=True)
    plt.title("Volume Distribution")
    plt.show()
else:
    print("Column 'Vol' not found.")
```

```python
numeric_df = df.select_dtypes(include=[np.number])

plt.figure(figsize=(14,10))
sns.heatmap(numeric_df.corr(), cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.show()
```
